```python
 import csv

with open('trainingdata.csv') as csvfile:
    data = list(csv.reader(csvfile))

print("Training Data:")
for row in data:
    print(row)

hypothesis = ['0'] * (len(data[0]) - 1)

for row in data:
    if row[-1].lower() == "yes":
        hypothesis = row[:-1]
        break

for row in data:
    if row[-1].lower() == "yes":
        for i in range(len(hypothesis)):
            if hypothesis[i] != row[i]:
                hypothesis[i] = '?'

print("\nMost Specific Hypothesis:")
print(hypothesis)
```
